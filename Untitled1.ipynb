```python
import numpy as np
marks=np.array([65,72,81,55,90])
print("Mean:",marks.mean())
print("Maximun:",marks.max())
print("Minimum:",marks.min())
print("Std Dev:",marks.std())
```

```python
import pandas as pd
data={
    "Name":["A","B","C","D","E"],
    "Marks":[65,72,81,55,90]
}
df=pd.DataFrame(data)
print(df)
print(df["Marks"].mean())
```

```python
import matplotlib.pyplot as plt
data={
    "Name":["A","B","C","D","E"],
    "Marks":[65,72,81,55,90]
}
plt.bar(df["Name"],df["Marks"])
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
plt.hist(df["Marks"])
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Marks")
plt.show()
```

```python
import numpy as np
result=np.random.choice(["Head","Tail"],size=10)
print(result)
```

```python
result=np.random.choice(["Head","Tail"],size=100)
heads=np.sum(result=="Head")
print("Heads:",heads)
print("Probability of Head:",heads/100)
```

```python
counts = [
    np.sum(result == "Head"),
    np.sum(result == "Tail")
]



plt.bar(["Head", "Tail"], counts)
plt.xlabel("Outcome")
plt.ylabel("Count")
plt.title("Coin Toss Simulation")
plt.show()
```

```python
result=np.random.choice(["1","2","3","4","5","6"],size=10)
print(result)
for i in range(1,7):
  h1=np.sum(result==str(i))
  print(i,":",h1)
  print("Probability of ",i,":",h1/10)
```
