```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
```

```python
#Normal distribution
x=np.linspace(-4,4,400)
#create
std_normal=stats.norm(0,1)
print('68-95-99.7 Rule:')
print(f'P(-1<z<1)={std_normal.cdf(1)-std_normal.cdf(-1):.6f}')
print(f'P(-2<z<2)={std_normal.cdf(2)-std_normal.cdf(-2):.6f}')
print(f'P(-3<z<3)={std_normal.cdf(3)-std_normal.cdf(-3):.6f}')
```

```python
print(f'P(-10<z<10)={std_normal.cdf(10)-std_normal.cdf(-10):.6f}')
```

```python
#z-score standardisation
mu_exam,sigma_exam=70,10
for score in [60,70,80,90]:
  z=(score - mu_exam)/sigma_exam
  pct=std_normal.cdf(z)*100
  print(f'Z-score for {score} is {z:.2f},Percentile={pct:.1f}%')
```

```python
#exponential distribution
lam=1.0
exp_dist=stats.expon(scale=1/lam)
print(f'Exp(lamb=1):Mean={exp_dist.mean():.4f},Variance={exp_dist.var():.4f}')
```

```python
#memory property
print(f'P(x>2)={1-exp_dist.cdf(2):.6f}')
print(f'P(x>3|x>1)={(1-exp_dist.cdf(3))/(1-exp_dist.cdf(1)):.6f}')
```

```python
#clt central limit theorem
#some amt of data follows diff distribution
#how can we relate the means of diff distribution
#we apply various distrubtion and consider its mean
#those means are distributed normally
sample_sizes={200,500,30,100}
fig,axes=plt.subplots(1,4,figsize=(16,4))
for i,n in enumerate(sample_sizes):
  #create random
  means=[np.mean(np.random.exponential(1,n)) for _ in range(3000)]
  axes[i].hist(means,bins=50,density=True,color='steelblue',edgecolor='black',alpha=0.7)
  axes[i].set_title(f'n={n}\nMean={np.mean(means):.4f}')
  axes[i].set_xlabel('Sample Mean')

fig.suptitle('CLT:Sample Mean of Exp(1)',fontsize=13,fontweight='bold')
plt.tight_layout();plt.show()
```
