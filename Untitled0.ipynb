```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
np.random.seed(42)
```

```python
n,p=20,0.35
binom=stats.binom(n,p)
x=np.arange(0,n+1)
```

```python
print(f'Mean={binom.mean():.4f}(Theory: np={np})')
```

```python
print("variance:",binom.var())
print("E(x2):",binom.moment(2))
print("Skewness:",binom.stats("s"))
print("CDF:",binom.cdf(10))
print("PMF:",binom.pmf(7))
```

```python
for lam in [1,3,5,10]:
  d=stats.poisson(lam)
  sim=np.random.poisson(lam,5000)
print("SIM:",sim)
```

```python
print("Mean:",d.mean())
print("Variance:",d.var())
print("E(x2):",d.moment(2))
print("Skewness:",d.stats("s"))
print("CDF:",d.cdf(10))
print("PMF:",d.pmf(7))
sim=np.random.poisson(lam,5000)
print("sim:",sim)
```

```python
p_geo=0.3
geo=stats.geom(p_geo)
print("Geo:",geo)
print("MEan:",geo.mean())
print("Variance:",geo.var())
print("E(x2):",geo.moment(2))
print("Skewness:",geo.stats("s"))
print("CDF:",1-geo.cdf(5)) #greater than 5
print("PMF:",geo.pmf(1))
```
